## 4. Explore the Database

Before starting the analysis, the database structure and available indicators are explored.

```python
pd.read_sql("SELECT DISTINCT IndicatorName FROM Indicators WHERE IndicatorName LIKE 'GDP%';", conn)
```

```python
gdp = pd.read_sql("""
SELECT CountryName, Year, Value
FROM Indicators
WHERE IndicatorName = 'GDP at market prices (current US$)'
AND CountryCode IN (
    'EGY', 'IRN', 'IRQ', 'LBY', 'MAR',
    'SAU', 'TUN', 'ARE', 'WBG', 'YEM', 'SYR'
)""", conn)

gdp.head()
```

```python
growth = pd.read_sql("""
SELECT CountryName, Year, Value
FROM Indicators
WHERE IndicatorName = 'GDP growth (annual %)'
AND CountryCode IN (
    'EGY', 'IRN', 'IRQ', 'LBY', 'MAR',
    'SAU', 'TUN', 'ARE', 'WBG', 'YEM', 'SYR'
)""", conn)

growth.head()
```

```python
inflation = pd.read_sql("""
SELECT CountryName, Year, Value
FROM Indicators
WHERE IndicatorName = 'Inflation, consumer prices (annual %)'
AND CountryCode IN (
    'EGY', 'IRN', 'IRQ', 'LBY', 'MAR',
    'SAU', 'TUN', 'ARE', 'WBG', 'YEM', 'SYR'
)""", conn)

inflation.head()
```

```python
growth = growth.sort_values("Year")

growth.head()
```

### GDP Growth Over Time

The line chart shows how GDP growth (annual %) changed over time across the selected countries.

```python
plt.figure(figsize=(18, 9))

for country in growth["CountryName"].unique():
    country_data = growth[growth["CountryName"] == country]

    plt.plot(
        country_data["Year"],
        country_data["Value"],
        label=country,
        linewidth=1.8
    )

plt.xlabel("Year", fontsize=13)
plt.ylabel("GDP Growth (%)", fontsize=13)

plt.title(
    "GDP Growth Over Time",
    fontsize=18,
    fontweight="bold"
)

plt.grid(True, alpha=0.3)

plt.legend(
    title="Country",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    fontsize=10,
    title_fontsize=11
)

plt.tight_layout()
plt.show()
```

```python
gdp_2014 = gdp[gdp["Year"] == 2014]

gdp_2014
```

### GDP Comparison (2014)

The bar chart compares GDP (current US$) across the selected countries in 2014.

```python
plt.figure(figsize=(16, 8))

plt.bar(
    gdp_2014["CountryName"],
    gdp_2014["Value"]
)

plt.xlabel("Country", fontsize=13)
plt.ylabel("GDP (Current US$)", fontsize=13)

plt.title(
    "GDP Comparison in 2014",
    fontsize=18,
    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right",
    fontsize=10
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python
inflation.groupby("CountryName")["Value"].mean()
```

```python
inflation = inflation.sort_values("Year")

inflation.head()
```

### Average Inflation by Country

The bar chart shows the average inflation rate (%) for each selected country across all available years.

```python
average_inflation = inflation.groupby("CountryName")["Value"].mean()

plt.figure(figsize=(16, 8))

plt.bar(
    average_inflation.index,
    average_inflation.values
)

plt.xlabel("Country", fontsize=13)
plt.ylabel("Inflation (%)", fontsize=13)

plt.title(
    "Average Inflation",
    fontsize=18,
    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right",
    fontsize=10
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```
